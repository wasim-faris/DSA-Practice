```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        

a = Node(20)
b= Node(30)
c = Node(40)

a.next = b
b.next = c

new_node = Node(44)
new_node.next = a
a = new_node

temp = a

while temp:
    print(temp.data, end=" -> ")
    temp = temp.next
print("None")
```
