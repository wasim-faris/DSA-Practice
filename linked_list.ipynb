```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


def add_node(head, value):
    temp = head

    # Check duplicate
    while temp:
        if temp.data == value:
            print("Duplicate value")
            return

        temp = temp.next

    # Move to last node
    temp = head
    while temp.next:
        temp = temp.next

    # Add new node
    temp.next = Node(value)


# Create linked list
a = Node(10)
b = Node(20)
c = Node(30)

a.next = b
b.next = c

head = a

add_node(head, 40)
add_node(head, 20)

# Print list
temp = head
while temp:
    print(temp.data, end=" -> ")
    temp = temp.next

print("None")
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

a = Node(23)
b = Node(345)
c = Node(123)
d = Node(2311)
e = Node(1234)

a.next = b
b.next = c
c.next = d
d.next = e

target = 23543

head = a 
temp = head 

found = False 

while temp:
    if temp.data == target:
        found = True
        break
    else:
        found = False
    temp = temp.next
        
if found:
    print("target founded")
else:
    print("Target not founded")
```
