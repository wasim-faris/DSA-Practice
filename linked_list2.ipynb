```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
a = Node(45)
b = Node(23)
c = Node(55)
d = Node(354)

a.next = b 
b.next = c
c.next = d

head = a 

temp = head
dup = False
new_value = 32

while temp:
    if temp.data == new_value:
        dup = True
        break
    temp = temp.next
    
if dup:
    print("You cant add the new value because it is already exists")
else:
    temp = head
    
    while temp.next:
        temp = temp.next
        
    temp.next = Node(new_value)


temp = head 

while temp:
    print(temp.data, end="->")
    temp = temp.next
print("None")
```
